```python
import os
import time
import soundfile as sf
import torch
from tqdm import tqdm

# -----------------------------
# paths
# -----------------------------
SAVE_DIR = "/content/drive/MyDrive/tts_benchmark/f5tts_wavs"
os.makedirs(SAVE_DIR, exist_ok=True)

# -----------------------------
# metrics
# -----------------------------
total_tts_time = 0
total_audio_duration = 0

audio_paths = []

# -----------------------------
# inference loop
# -----------------------------

start_all = time.time()

for idx, row in tqdm(df.iterrows(), total=len(df)):

    text = row["sentence"]  # or "text" depending on your dataset

    audio_path = os.path.join(SAVE_DIR, f"{idx}.wav")

    start = time.time()

    # -----------------------------
    # F5TTS inference
    # -----------------------------
    out = model.infer(
        ref_file=row["audio_path"],   # keep original reference audio
        ref_text=row["sentence"],     # reference text (or separate ref column if you have it)
        gen_text=text,
    )

    # -----------------------------
    # extract waveform safely
    # -----------------------------
    # F5TTS returns tuple:
    # (waveform, sr, mel/spec)
    if isinstance(out, tuple):
        wav = out[0]
        sr = out[1] if len(out) > 1 else 24000
    else:
        wav = out
        sr = 24000

    wav = torch.tensor(wav)

    # -----------------------------
    # ensure correct shape
    # -----------------------------
    if wav.ndim > 1:
        wav = wav.squeeze()

    wav = wav.detach().cpu().numpy()

    # -----------------------------
    # save audio
    # -----------------------------
    sf.write(audio_path, wav, sr)

    # -----------------------------
    # timing
    # -----------------------------
    tts_time = time.time() - start
    total_tts_time += tts_time

    duration = len(wav) / sr
    total_audio_duration += duration

    audio_paths.append(audio_path)

end_all = time.time()

# -----------------------------
# RTF
# -----------------------------
rtf = total_tts_time / total_audio_duration

print("\n================ TTS RUNTIME =================")
print(f"Samples: {len(audio_paths)}")
print(f"Total TTS time: {total_tts_time:.2f} sec")
print(f"Generated audio duration: {total_audio_duration:.2f} sec")
print(f"RTF: {rtf:.4f}")
print("==============================================\n")
```

```python
import os
import time
import torch
import soundfile as sf
from tqdm import tqdm

# -----------------------------
# paths
# -----------------------------
SAVE_DIR = "/content/drive/MyDrive/tts_benchmark/f5tts_wavs"
os.makedirs(SAVE_DIR, exist_ok=True)

CHECKPOINT_PATH = os.path.join(SAVE_DIR, "progress.txt")

# -----------------------------
# load resume index
# -----------------------------
if os.path.exists(CHECKPOINT_PATH):
    start_idx = 937
else:
    start_idx = 937

print(f"Resuming from index: {start_idx}")

# -----------------------------
# metrics
# -----------------------------
total_tts_time = 0
total_audio_duration = 0

# -----------------------------
# IMPORTANT: do NOT call model.eval()
# -----------------------------

for idx, row in tqdm(df.iterrows(), total=len(df)):

    # skip already processed
    if idx < start_idx:
        continue

    text = row["sentence"]
    audio_path = os.path.join(SAVE_DIR, f"{idx}.wav")

    try:
        start = time.time()

        out = model.infer(
            ref_file=row["audio_path"],
            ref_text=row["sentence"],
            gen_text=text
        )

        # -------------------------
        # extract waveform safely
        # -------------------------
        if isinstance(out, tuple):
            wav, sr = out[0], out[1]
        else:
            wav, sr = out, 24000

        wav = torch.tensor(wav)

        if wav.ndim > 1:
            wav = wav.squeeze()

        wav = wav.detach().cpu().numpy()

        # -------------------------
        # save wav
        # -------------------------
        sf.write(audio_path, wav, sr)

        # -------------------------
        # timing
        # -------------------------
        tts_time = time.time() - start
        total_tts_time += tts_time
        total_audio_duration += len(wav) / sr

        # -------------------------
        # checkpoint update (CRITICAL)
        # -------------------------
        with open(CHECKPOINT_PATH, "w") as f:
            f.write(str(idx + 1))

    except Exception as e:
        print(f"FAILED at {idx}: {e}")
        continue

# -----------------------------
# final RTF
# -----------------------------
rtf = total_tts_time / total_audio_duration

print("\n===== RESUME DONE =====")
print(f"Processed up to: {idx}")
print(f"RTF (partial run): {rtf:.4f}")
```

```python
import os
import time
import soundfile as sf
import torch
import pandas as pd
from tqdm import tqdm
import csv

# -----------------------------
# PATHS
# -----------------------------
SAVE_DIR = "/content/drive/MyDrive/tts_benchmark/f5tts_wavs"
CSV_PATH = "/content/drive/MyDrive/tts_benchmark/f5tts_metadata.csv"

os.makedirs(SAVE_DIR, exist_ok=True)

# -----------------------------
# LOAD DATA
# -----------------------------
# df must contain:
# - sentence (text)
# - audio_path (reference audio for F5TTS)
# -----------------------------

# df = ...

# -----------------------------
# RESUME LOGIC
# -----------------------------
def get_resume_index(csv_path):
    if not os.path.exists(csv_path):
        return 0
    try:
        df_old = pd.read_csv(csv_path)
        return len(df_old)
    except:
        return 0

start_idx = get_resume_index(CSV_PATH)
print(f"\nResuming from index: {start_idx}\n")

# -----------------------------
# CREATE CSV HEADER IF NEEDED
# -----------------------------
if not os.path.exists(CSV_PATH):
    with open(CSV_PATH, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow([
            "idx",
            "audio_path",
            "text",
            "tts_runtime",
            "audio_duration",
            "tts_rtf"
        ])

# -----------------------------
# MAIN LOOP
# -----------------------------
for idx, row in tqdm(df.iloc[start_idx:].iterrows(),
                     total=len(df) - start_idx):

    text = row["sentence"]
    ref_audio = row["audio_path"]

    audio_path = os.path.join(SAVE_DIR, f"{idx}.wav")

    try:
        start = time.time()

        # -----------------------------
        # F5TTS inference
        # -----------------------------
        out = model.infer(
            ref_file=ref_audio,
            ref_text=row["sentence"],   # change if you have separate ref_text column
            gen_text=text,
        )

        # -----------------------------
        # extract waveform
        # -----------------------------
        if isinstance(out, tuple):
            wav = out[0]
            sr = out[1] if len(out) > 1 else 24000
        else:
            wav = out
            sr = 24000

        wav = torch.tensor(wav)

        if wav.ndim > 1:
            wav = wav.squeeze()

        wav = wav.detach().cpu().numpy()

        # -----------------------------
        # SAVE AUDIO IMMEDIATELY
        # -----------------------------
        sf.write(audio_path, wav, sr)

        # -----------------------------
        # METRICS
        # -----------------------------
        tts_time = time.time() - start
        duration = len(wav) / sr
        tts_rtf = tts_time / duration if duration > 0 else 0

        # -----------------------------
        # APPEND CSV (CRASH SAFE)
        # -----------------------------
        with open(CSV_PATH, "a", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow([
                idx,
                audio_path,
                text,
                tts_time,
                duration,
                tts_rtf
            ])
            f.flush()

        print(f"[{idx}] OK | RTF={tts_rtf:.4f}")

    except Exception as e:
        print(f"[{idx}] FAILED:", e)
        continue
```

```python
import pandas as pd

df_res = pd.read_csv(CSV_PATH)

summary = {
    "model": "F5TTS",
    "samples": len(df_res),
    "tts_rtf": df_res["tts_rtf"].mean(),
    "tts_latency_sec": df_res["tts_runtime"].mean(),
    "aggregate_rtf": df_res["tts_runtime"].sum() / df_res["audio_duration"].sum(),
    "speedup": 1 / df_res["tts_rtf"].mean()
}

summary_df = pd.DataFrame([summary])
summary_df.to_csv("/content/drive/MyDrive/tts_benchmark/f5tts_summary.csv", index=False)

print(summary_df)
```
