```python
rows = []

for file_path in tqdm(audio_files):

    waveform, sr = librosa.load(file_path, sr=16000)

    waveform = torch.tensor(waveform).unsqueeze(0).to(device)

    with torch.no_grad():
        mos_pred = nisqa(waveform)

    # FIX: handle multi-output tensor
    mos_pred = mos_pred.squeeze().detach().cpu().numpy()

    rows.append({
        "file": os.path.basename(file_path),

        "mos_pred": float(mos_pred[0]),
        "noi_pred": float(mos_pred[1]),
        "dis_pred": float(mos_pred[2]),
        "col_pred": float(mos_pred[3]),
        "loud_pred": float(mos_pred[4]),
    })
```

###################################################### F5-TTS

```python
import os
import re
import torch
import pandas as pd
import librosa
from tqdm import tqdm
from torchmetrics.audio import NonIntrusiveSpeechQualityAssessment
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

nisqa = NonIntrusiveSpeechQualityAssessment(
    fs=16000
).to(device)
```

```python
METADATA_PATH = "/content/drive/MyDrive/tts_benchmark/f5tts_metadata.csv"
df = pd.read_csv(
    METADATA_PATH
)

df.head()
```

```python
audio_files = df["audio_path"].tolist()
```

```python
import torch
import torch.nn.functional as F

MIN_LENGTH = 48000  # ~3 sec at 16kHz

results = []

for _, row in tqdm(df.iterrows(), total=len(df)):

    wav_path = row["audio_path"]

    waveform, sr = librosa.load(
        wav_path,
        sr=16000,
        mono=True
    )

    waveform = torch.tensor(
        waveform,
        dtype=torch.float32
    ).unsqueeze(0).to(device)

    # pad short files
    if waveform.shape[1] < MIN_LENGTH:

        pad_amount = MIN_LENGTH - waveform.shape[1]

        waveform = F.pad(
            waveform,
            (0, pad_amount),
            mode="constant",
            value=0
        )

    with torch.no_grad():
        scores = nisqa(waveform)

    scores = scores.squeeze().cpu().numpy()

    results.append({
        "audio_path": wav_path,
        "mos_pred": scores[0],
        "noi_pred": scores[1],
        "dis_pred": scores[2],
        "col_pred": scores[3],
        "loud_pred": scores[4],
    })

results_df = pd.DataFrame(results)
```

```python
results_df = pd.DataFrame(results)

print(results_df.head())
print("Samples:", len(results_df))
```

```python
summary = pd.DataFrame([{
    "model": "F5TTS",
    "samples": len(results_df),

    "nisqa_mos_mean": results_df["mos_pred"].mean(),
    "nisqa_mos_std": results_df["mos_pred"].std(),
    "nisqa_mos_min": results_df["mos_pred"].min(),
    "nisqa_mos_max": results_df["mos_pred"].max(),

    "noise_mean": results_df["noi_pred"].mean(),
    "noise_std": results_df["noi_pred"].std(),

    "discontinuity_mean": results_df["dis_pred"].mean(),
    "discontinuity_std": results_df["dis_pred"].std(),

    "coloration_mean": results_df["col_pred"].mean(),
    "coloration_std": results_df["col_pred"].std(),

    "loudness_mean": results_df["loud_pred"].mean(),
    "loudness_std": results_df["loud_pred"].std(),
}])

summary
```

```python
import os

SAVE_DIR = "/content/drive/MyDrive/tts_benchmark/F5TTS"
os.makedirs(SAVE_DIR, exist_ok=True)
```

```python
DETAIL_PATH = os.path.join(
    SAVE_DIR,
    "F5TTS_nisqa_scores.csv"
)

results_df.to_csv(
    DETAIL_PATH,
    index=False
)

print("Saved detailed scores:")
print(DETAIL_PATH)
```

```python
SUMMARY_PATH = os.path.join(
    SAVE_DIR,
    "F5TTS_nisqa_summary.csv"
)

summary.to_csv(
    SUMMARY_PATH,
    index=False
)

print("Saved summary:")
print(SUMMARY_PATH)
```
